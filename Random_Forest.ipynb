#Random Forest


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn import datasets
```

```python
iris = datasets.load_iris()
```

```python
data = pd.DataFrame({"sepal_length": iris.data[:, 0],
                     "sepal_width": iris.data[:, 1],
                     "petal_length": iris.data[:, 2],
                     "petal_width": iris.data[:, 3],
                     "ouput":iris.target})
```

```python
data
```

```python
#split the dataset in train-test split
```

```python
from sklearn.model_selection import train_test_split
```

```python
#featuers and output spliting
x=data[['sepal_length','sepal_width','petal_length','petal_width']]
y=data['ouput']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20)
```

```python
#model define
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(n_estimators=100)
```

```python
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
#classsfiction_score
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt
```

```python
from sklearn.tree import plot_tree
plot_tree(model.estimators_[0], filled=True, feature_names=['sepal_length','sepal_width','petal_length','petal_width'])
plt.show()
```
